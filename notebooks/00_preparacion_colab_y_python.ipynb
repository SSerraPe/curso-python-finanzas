## 1. ¿Qué es esto?

Un **notebook** es un documento que mezcla **texto** (como este) y **celdas de código** que puedes ejecutar. Es como una hoja de Excel en la que, en lugar de fórmulas escondidas en las celdas, el cálculo está escrito y se lee de arriba abajo.

**Google Colab** ejecuta el notebook en un servidor de Google, así que no tienes que instalar nada.

### Guarda tu copia

## 2. Python como calculadora

Python entiende operaciones como Excel: `+ - * /`. La potencia es `**` (en Excel `^`).

```python
1250 * 1.21        # precio con IVA
```

```python
(100 - 72) / 100   # margen en tanto por uno
```

```python
1.05 ** 3          # crecimiento compuesto 3 años al 5%
```

Lo que va detrás de `#` es un **comentario**: Python lo ignora. Úsalo para explicar *por qué* haces algo.

## 3. Variables: ponerle nombre a las cosas

En Excel, el dato vive en la celda `B7` y hay que recordar qué era `B7`. En Python le damos un **nombre**:

```python
prima = 480.0          # ingresos por la póliza (€)
coste_siniestros = 350.0
comision = prima * 0.15

margen = prima - coste_siniestros - comision
margen
```

Si ahora cambias `prima` y vuelves a ejecutar las celdas, el margen se recalcula. Es lo mismo que cambiar un input en Excel.

**Reglas para los nombres:** minúsculas, palabras separadas por `_`, sin tildes ni espacios, y que expliquen lo que contienen: `prima_devengada` es mejor que `x` o `pd2`.

## 4. Texto y f-strings

El texto (*string*) va entre comillas. Para mezclar texto y números se usan las **f-strings**: una `f` delante y las variables entre `{}`.

```python
canal = "Agentes"
print(f"El canal {canal} tiene un margen de {margen} €")
print(f"Con formato de miles y 2 decimales: {margen:,.2f} €")
print(f"Como porcentaje: {margen / prima:.1%}")
```

Ejecuta la celda de abajo. Va a dar un **error** a propósito.

```python
# Este error es intencionado: 'primas' (con s) no existe, la variable se llama 'prima'
try:
    print(primas)
except NameError as e:
    print("NameError:", e)
```

En la vida real verás un recuadro rojo con el texto `NameError: name 'primas' is not defined`. Cómo leerlo:

1. **La última línea es la importante**: dice el *tipo* de error y el motivo.
2. Encima se indica la línea donde ha fallado.
3. Los errores más típicos al empezar son `NameError` (nombre mal escrito), `SyntaxError` (falta un paréntesis o unas comillas), `TypeError` (sumar texto con números) e `IndentationError` (espacios al principio de la línea).
