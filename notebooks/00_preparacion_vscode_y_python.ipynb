## 1. ¿Qué es esto?

Un **notebook** es un documento que mezcla **texto** (como este) y **celdas de código** que puedes ejecutar. Es como una hoja de Excel en la que, en lugar de fórmulas escondidas en las celdas, el cálculo está escrito y se lee de arriba abajo.

**VS Code** es el programa con el que abrimos y ejecutamos los notebooks en tu ordenador. Python corre en tu máquina, dentro del entorno **`.venv`** del curso, que tiene instaladas las librerías que vamos a usar.

**La carpeta `curso-python-finanzas`** es tu copia del curso: notebooks, datos (`data/`), funciones (`src/`) y lo que generes (`output/`). **GitHub** es el original del profesor, de solo lectura. Tu carpeta es tu cuaderno: lo que cambies se queda en tu ordenador.

## 4. Python como calculadora

Python entiende operaciones como Excel: `+ - * /`. La potencia es `**` (en Excel `^`).

```python
1250 * 1.21        # precio con IVA
```

```python
(100 - 72) / 100   # margen en tanto por uno
```

```python
1.05 ** 3          # crecimiento compuesto 3 años al 5%
```

Lo que va detrás de `#` es un **comentario**: Python lo ignora. Úsalo para explicar *por qué* haces algo.

## 5. Variables: ponerle nombre a las cosas

En Excel, el dato vive en la celda `B7` y hay que recordar qué era `B7`. En Python le damos un **nombre**:

```python
prima = 480.0          # ingresos por la póliza (€)
coste_siniestros = 350.0
comision = prima * 0.15

margen = prima - coste_siniestros - comision
margen
```

Si ahora cambias `prima` y vuelves a ejecutar las celdas, el margen se recalcula. Es lo mismo que cambiar un input en Excel.

**Reglas para los nombres:** minúsculas, palabras separadas por `_`, sin tildes ni espacios, y que expliquen lo que contienen: `prima_devengada` es mejor que `x` o `pd2`.

## 6. Texto y f-strings

El texto (*string*) va entre comillas. Para mezclar texto y números se usan las **f-strings**: una `f` delante y las variables entre `{}`.

```python
canal = "Agentes"
print(f"El canal {canal} tiene un margen de {margen} €")
print(f"Con formato de miles y 2 decimales: {margen:,.2f} €")
print(f"Como porcentaje: {margen / prima:.1%}")
```

## 9. 🔄 Solo si el profesor anuncia una actualización del curso

Si corregimos algo durante el curso, hay que traer la última versión **sin tocar tus notebooks**. No hace falta hacerlo ahora.

**Si descargaste el curso con Git** (`git clone`): en la terminal de VS Code (*Terminal → New Terminal*), ejecuta `git pull`. Si Git protesta porque has cambiado notebooks, avisa al profesor.

**Si lo descargaste como ZIP:**
1. Descarga otra vez el ZIP desde GitHub (*Code → Download ZIP*) y descomprímelo (por ejemplo, en *Descargas*).
2. Escribe abajo la ruta de la carpeta descomprimida y ejecuta la celda. Copia en tu curso los datos, `src/`, scripts, documentos y soluciones, y añade los notebooks nuevos que no tengas. **Tus notebooks no se tocan.**

```python
# 🔄 Actualizar mi copia del curso desde un ZIP descomprimido
import shutil
from pathlib import Path

CARPETA_NUEVA = Path("~/Downloads/curso-python-finanzas-main").expanduser()   # ← cambia la ruta si hace falta
SE_ACTUALIZAN = ["config", "data", "docs", "herramientas", "scripts", "src", "notebooks/soluciones", ".vscode",
                 "README.md", "requirements.txt", "requirements-colab.txt"]

mi_curso = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data").is_dir() and (p / "src").is_dir()), None)
if not (CARPETA_NUEVA / "notebooks").is_dir():
    print(f"ℹ️ No encuentro la versión nueva en {CARPETA_NUEVA}. Si no hay actualización, no hace falta ejecutar esta celda.")
elif mi_curso is None or CARPETA_NUEVA.resolve() == mi_curso.resolve():
    print("❌ La ruta debe ser la carpeta NUEVA descomprimida, no la de tu curso.")
else:
    for ruta in SE_ACTUALIZAN:
        origen, destino = CARPETA_NUEVA / ruta, mi_curso / ruta
        if origen.is_dir():
            shutil.copytree(origen, destino, dirs_exist_ok=True)
        elif origen.exists():
            shutil.copy2(origen, destino)
    nuevos = [nb.name for nb in (CARPETA_NUEVA / "notebooks").glob("*.ipynb")
              if not (mi_curso / "notebooks" / nb.name).exists()]
    for nombre in nuevos:
        shutil.copy2(CARPETA_NUEVA / "notebooks" / nombre, mi_curso / "notebooks" / nombre)
    print("✅ Copia actualizada. Tus notebooks no se han tocado.")
    print("   Notebooks nuevos:", ", ".join(nuevos) if nuevos else "ninguno")
    print("   Si ha cambiado requirements.txt, en la terminal:  pip install -r requirements.txt")
```
