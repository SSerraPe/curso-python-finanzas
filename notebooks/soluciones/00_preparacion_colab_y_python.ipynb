```python
1250 * 1.21        # precio con IVA
```

```python
(100 - 72) / 100   # margen en tanto por uno
```

## 5. Variables: ponerle nombre a las cosas

En Excel, el dato vive en la celda `B7` y hay que recordar qué era `B7`. En Python le damos un **nombre**:

```python
prima = 480.0          # ingresos por la póliza (€)
coste_siniestros = 350.0
comision = prima * 0.15

margen = prima - coste_siniestros - comision
margen
```

Si ahora cambias `prima` y vuelves a ejecutar las celdas, el margen se recalcula. Es lo mismo que cambiar un input en Excel.

**Reglas para los nombres:** minúsculas, palabras separadas por `_`, sin tildes ni espacios, y que expliquen lo que contienen: `prima_devengada` es mejor que `x` o `pd2`.

## 6. Texto y f-strings

El texto (*string*) va entre comillas. Para mezclar texto y números se usan las **f-strings**: una `f` delante y las variables entre `{}`.

```python
canal = "Agentes"
print(f"El canal {canal} tiene un margen de {margen} €")
print(f"Con formato de miles y 2 decimales: {margen:,.2f} €")
print(f"Como porcentaje: {margen / prima:.1%}")
```

## 7. Los errores son tus amigos

Ejecuta la celda de abajo. Va a dar un **error** a propósito.

```python
# Este error es intencionado: 'primas' (con s) no existe, la variable se llama 'prima'
try:
    print(primas)
except NameError as e:
    print("NameError:", e)
```

En la vida real verás un recuadro rojo con el texto `NameError: name 'primas' is not defined`. Cómo leerlo:

1. **La última línea es la importante**: dice el *tipo* de error y el motivo.
2. Encima se indica la línea donde ha fallado.
3. Los errores más típicos al empezar son `NameError` (nombre mal escrito), `SyntaxError` (falta un paréntesis o unas comillas), `TypeError` (sumar texto con números) e `IndentationError` (espacios al principio de la línea).

## 9. 🔄 Solo si el profesor anuncia una actualización del curso

Si corregimos algo durante el curso, esta celda descarga la última versión y actualiza tu copia **sin tocar tus notebooks**: renueva los datos, las funciones de `src/` y las soluciones, y añade los notebooks nuevos que no tengas.

No hace falta ejecutarla ahora.

```python
# @title 🔄 Actualizar mi copia del curso { display-mode: "form" }
import shutil, subprocess, sys
from pathlib import Path

REPO_URL = "https://github.com/SSerraPe/curso-python-finanzas.git"
COPIA_DRIVE = Path("/content/drive/MyDrive/curso-python-finanzas")
SE_ACTUALIZAN = ["config", "data", "docs", "herramientas", "scripts", "src", "notebooks/soluciones",
                 "README.md", "requirements.txt", "requirements-colab.txt"]

if "google.colab" not in sys.modules:
    print("ℹ️ No estás en Colab: en local, actualiza con `git pull`.")
else:
    from google.colab import drive
    drive.mount("/content/drive")
    if not COPIA_DRIVE.exists():
        print("❌ No encuentro tu copia en Mi unidad › curso-python-finanzas. Ejecuta antes la celda 📥 de la sección 2.")
    else:
        tmp = Path("/content/_descarga_curso")
        shutil.rmtree(tmp, ignore_errors=True)
        subprocess.run(["git", "clone", "-q", "--depth", "1", REPO_URL, str(tmp)], check=True)
        for ruta in SE_ACTUALIZAN:
            origen, destino = tmp / ruta, COPIA_DRIVE / ruta
            if origen.is_dir():
                shutil.copytree(origen, destino, dirs_exist_ok=True)
            elif origen.exists():
                shutil.copy2(origen, destino)
        nuevos = [nb.name for nb in (tmp / "notebooks").glob("*.ipynb")
                  if not (COPIA_DRIVE / "notebooks" / nb.name).exists()]
        for nombre in nuevos:
            shutil.copy2(tmp / "notebooks" / nombre, COPIA_DRIVE / "notebooks" / nombre)
        drive.flush_and_unmount()
        print("✅ Copia actualizada. Tus notebooks no se han tocado.")
        print("   Notebooks nuevos:", ", ".join(nuevos) if nuevos else "ninguno")
```
